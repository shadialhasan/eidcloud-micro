```python
# [Cell 2] Test REST API Endpoints with Python & Curl
import json
import urllib.request

BASE_URL = "http://127.0.0.1:8080"

# 1. Health check & Telemetry
req = urllib.request.urlopen(f"{BASE_URL}/health")
print("=== GET /health ===")
print(f"Status: {req.status}")
print(f"X-Response-Time: {req.headers.get('X-Response-Time')}")
print(json.dumps(json.loads(req.read().decode()), indent=2))

# 2. Create User (POST with JSON payload)
user_payload = json.dumps({"name": "Dr. Tariq Zaid", "role": "Cloud Architect"}).encode('utf-8')
post_req = urllib.request.Request(
    f"{BASE_URL}/api/users",
    data=user_payload,
    headers={"Content-Type": "application/json"},
    method="POST"
)
with urllib.request.urlopen(post_req) as resp:
    print("\n=== POST /api/users ===")
    print(f"Status: {resp.status}")
    print(json.dumps(json.loads(resp.read().decode()), indent=2))

# 3. Retrieve User by Dynamic Parameter
user_req = urllib.request.urlopen(f"{BASE_URL}/api/users/1")
print("\n=== GET /api/users/1 ===")
print(json.dumps(json.loads(user_req.read().decode()), indent=2))
```
